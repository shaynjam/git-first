```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
dataset=pd.read_csv(r"C:\Users\shaya\Downloads\housing.csv.zip")
dataset=dataset[["median_house_value","median_income"]]
print(dataset["median_income"].isna().sum())
print(dataset["median_house_value"].isna().sum())
print(dataset.head(10))
```

```python
#dividing by 100k because median house value is numerically great in dataset so gradient descent will struggle with such a mismatch.
X=dataset["median_house_value"].to_numpy()/100000 
Y=dataset["median_income"].to_numpy() 
print(X,"\n",Y)
print(X.shape)
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
print(X_test.shape)
m=X_train.shape[0]
m
```

```python
def predictor(x,w,b):
    return (w*x + b)
```

```python
def Loss_Fn(x,y,w_new,b_new):
    y_hat=predictor(X_train,w_new,b_new)
    v=((y_hat-Y_train)**2)
    sum=np.sum(v)
    J=sum/(2*m)
    return J
Loss_Fn(x=X_train,y=Y_train,w_new=1,b_new=1)
```
